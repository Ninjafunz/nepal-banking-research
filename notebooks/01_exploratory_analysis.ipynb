# Nepal Commercial Banking — Empirical Exploratory Analysis
**Sample:** 23 Class A Commercial Banks (FY2020 – FY2025)
**Objective:** Quick exploratory visualization of sector concentration (HHI), profitability differentials (State-Owned vs. Private), and business model spreads.

```python
import os

import pandas as pd
import plotly.express as px

# Load Master Panel
panel_path = os.path.join('..', 'MASTER', 'master_bank_panel.csv')
df = pd.read_csv(panel_path)
print(f"Loaded Master Panel: {df.shape[0]} bank-years, {df.shape[1]} variables.")
df.head(3)
```

## 1. Industry Concentration Trends (HHI & Market Shares)

```python
ind_path = os.path.join('..', 'ANALYSIS', 'industry_structure.xlsx')
ind_df = pd.read_excel(ind_path, sheet_name='industry_structure')

fig_hhi = px.line(
    ind_df, x='fy', y=['hhi_assets', 'hhi_loans', 'hhi_deposits'],
    markers=True, title='Herfindahl-Hirschman Index (HHI) Trend (FY2020-FY2025)',
    labels={'value': 'HHI (0-10,000)', 'variable': 'Market Segment', 'fy': 'Fiscal Year'}
)
fig_hhi.show()
```

## 2. Ownership Structure & Profitability (State-Owned vs. Private Banks)

```python
df['ownership_type'] = df['state_owned'].map({True: 'State-Owned', False: 'Private'})

fig_roe = px.box(
    df, x='fy', y='roe', color='ownership_type',
    title='Return on Equity (ROE %) by Ownership Structure over Time',
    labels={'roe': 'Return on Equity (%)', 'fy': 'Fiscal Year'}
)
fig_roe.show()
```

## 3. Net Interest Margin (NIM) vs. Return on Assets (ROA)

```python
fig_scatter = px.scatter(
    df[df['fy'] == 2024], x='nim', y='roa', size='total_assets',
    color='bank_code', hover_name='bank_name',
    title='NIM vs. ROA in FY2024 (Bubble Size = Asset Scale)',
    labels={'nim': 'Net Interest Margin (%)', 'roa': 'Return on Assets (%)'}
)
fig_scatter.show()
```
